```python
# 1. Simple Greedy algorithm for a Multi-Armed Bandit problem
n = int(input("Enter number of machines: "))
machines = []
q_values = []

for i in range(n):
    name = input(f"Enter machine name {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards: ").split()))

    q = sum(rewards) / len(rewards)

    machines.append((name, rewards))
    q_values.append(q)

print("\nEstimated Q(a) Values:")

for i in range(n):
    name, rewards = machines[i]
    print(f"\nMachine {name}")
    print(f"Rewards: {[int(r) if r.is_integer() else r for r in rewards]}")
    print(f"Q({name}) = {q_values[i]:.2f}")

best_index = q_values.index(max(q_values))
best_machine = machines[best_index][0]

print(f"\nGreedy Selected Machine: {best_machine}")

# ========================================== #
# 2. Greedy algorithm for a Multi-Armed Bandit problem using the incremental update method
print("\n--- Running Incremental Update Method ---\n")
n_inc = int(input("Enter number of machines (Incremental): "))
machines_inc = []
q_values_inc = []
counts_inc = []

for i in range(n_inc):
    name = input(f"Enter machine name {i + 1}: ")
    rewards = list(map(float, input("Enter observed rewards: ").split()))
    machines_inc.append((name, rewards))
    q_values_inc.append(0.0)
    counts_inc.append(0)

for i in range(n_inc):
    name, rewards = machines_inc[i]
    print(f"\nMachine {name}")

    for reward in rewards:
        counts_inc[i] += 1
        q_values_inc[i] = q_values_inc[i] + (reward - q_values_inc[i]) / counts_inc[i]
        print(
            f"Reward = {int(reward) if reward.is_integer() else reward}, "
            f"N({name}) = {counts_inc[i]}, "
            f"Q({name}) = {q_values_inc[i]:.2f}"
        )

print("\nFinal Estimated Q(a) Values:")

for i in range(n_inc):
    name, rewards = machines_inc[i]
    formatted_rewards = [
        int(r) if r.is_integer() else r for r in rewards
    ]

    print(f"\nMachine {name}")
    print(f"Rewards: {formatted_rewards}")
    print(f"Q({name}) = {q_values_inc[i]:.2f}")

best_index_inc = q_values_inc.index(max(q_values_inc))
best_machine_inc = machines_inc[best_index_inc][0]

print(f"\nGreedy Selected Machine (Incremental): {best_machine_inc}")
```
